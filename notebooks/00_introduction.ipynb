# Analysing real data

We're now at a stage where you can analyse some real sequence data. We're going to read this in directly from a url: [https://raw.githubusercontent.com/liampshaw/Computation-for-Bioscientists/refs/heads/main/data/labelled_chunks.fasta](https://raw.githubusercontent.com/liampshaw/Computation-for-Bioscientists/refs/heads/main/data/labelled_chunks.fasta). If you click on that link you'll see what the file looks like. The function in the next cell allows us to read this in directly and save the result as a dictionary.

````python
import urllib

def read_fasta_from_url(url):
    with urllib.request.urlopen(url) as response:
        text = response.read().decode('utf-8')

    sequences = {}
    header = None
    seq_lines = []

    for line in text.splitlines():
        if line.startswith('>'):
            if header is not None:
                sequences[header] = ''.join(seq_lines)
            header = line[1:].strip()  # drop '>', keep rest of header line
            seq_lines = []
        else:
            seq_lines.append(line.strip())

    if header is not None:  # save the last record
        sequences[header] = ''.join(seq_lines)

    return sequences
````

````python
# Use the above function to read in from the url given
our_sequences = read_fasta_from_url('https://raw.githubusercontent.com/liampshaw/Computation-for-Bioscientists/refs/heads/main/data/labelled_chunks.fasta')
````

This has created a variable ```our_sequences``` that contains sequences from different organisms. We're going to use our ```count_kmers``` function to count k-mers in each of these sequences.

````python
# Count k-mers in our_sequences
kmer_counts = {}
for sequence_id in our_sequences.keys():
  kmer_counts[sequence_id] = count_kmers(our_sequences[sequence_id], 4)

# You might wish to look at kmer_counts by typing it on the next line in this cell before executing the cell
````
